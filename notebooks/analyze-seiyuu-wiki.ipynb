# Analyze Seiyuu in Wikipedia

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
# Import data
df_original_data = pd.read_csv('../data/jppersonwiki.csv')
```

```python
# Print information about the dataframe
df_original_data.info()
```

```python
# Overview of missing values per column
for name, values in df_original_data.items():
  print(f'Missing values in column {name}: ', df_original_data[name].isnull().sum())

# Total number of missing values in dataset
total_nan = df_original_data.isnull().sum().sum()
print(f"Total number of NaN values in the DataFrame: {total_nan}")
```

```python
# Keep all records that contain seiyuu (声優) as occupation
df_seiyuu_original = df_original_data[df_original_data['occupations'].str.contains('声優')]
```

```python
# Print information about the seiyuu dataframe
df_seiyuu_original.info()

# Print all columns in the dataframe
df_seiyuu_original.columns
```

```python
df_seiyuu = df_seiyuu_original.copy()

# Drop columns from the dataframe between specific ranges
df_seiyuu.drop(df_seiyuu.columns[np.r_[15:19, 21:36]], axis=1, inplace=True)
```

```python
# Print information about the new seiyuu dataframe
df_seiyuu.info()

# Print all columns in the dataframe
df_seiyuu.columns
```

```python
# Only fetch the birth year
df_seiyuu_filtered = df_seiyuu[['birth_year', 'gender']]

df_seiyuu_by_year = df_seiyuu_filtered.groupby(['birth_year']).size()
df_seiyuu_by_year_gender = df_seiyuu_filtered.groupby(['birth_year', 'gender']).size().unstack(fill_value=0)

# Drop column
df_seiyuu_by_year.drop(columns='トランス男性', inplace=True)
df_seiyuu_by_year_gender.drop(columns='トランス男性', inplace=True)
```

```python
df_seiyuu_by_year.index
```

```python
# Create a figure and axis object
fig, ax = plt.subplots()

# Create a plot
# ax.bar(df_seiyuu_by_year.index, df_seiyuu_by_year.values)
ax.plot(df_seiyuu_by_year.index, df_seiyuu_by_year_gender['M'], label='Male')
ax.plot(df_seiyuu_by_year.index, df_seiyuu_by_year_gender['F'], label='Female')

# set minimum y-axis value, title, labels, legend and grid
ax.set_ylim(bottom=0)
ax.set_title('Amount of seiyuu by birth year')
ax.set_xlabel('Birth year')
ax.set_ylabel('Count')
ax.legend()
ax.grid(linewidth=0.3)
```
